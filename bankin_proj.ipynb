```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

```

```python
bins = [0, 100000, 300000, float('inf')]         # Define 3 income ranges
labels = ["low", "mid", "high"]                  # Labels for each range

df['Income Band'] = pd.cut(df['Estimated Income'],
                           bins=bins,
                           labels=labels,
                           right=False)
```

```python
df['Income Band'].value_counts().plot(kind='bar')
```

```python
categorical_cols = df[['BRId',	"GenderId",	"IAId", "Amount of Credit Cards", "Nationality",	"Occupation",	"Fee Structure", "Loyalty Classification","Properties Owned"	, "Risk Weighting", "Income Band" ]].columns
```

```python
for cols in categorical_cols:
  print("cols:")
  display(df[cols].value_counts())
```

univariant analysis

```python
for i, predictor in enumerate(df[['BRId',	"GenderId",	"IAId", "Amount of Credit Cards", "Nationality",	"Occupation",	"Fee Structure", "Loyalty Classification","Properties Owned"	, "Risk Weighting", "Income Band" ]]):
    plt.figure(i)
    sns.countplot(data=df, x=predictor, color="blue")
```

Bivariant analysis

```python
for i, predictor in enumerate(df[['BRId',	"GenderId",	"IAId", "Amount of Credit Cards", "Nationality",	"Occupation",	"Fee Structure", "Loyalty Classification","Properties Owned"	, "Risk Weighting", "Income Band" ]]):
    plt.figure(i)
    sns.countplot(data=df, x=predictor,hue="GenderId", color="blue")
```

```python
for i, predictor in enumerate(df[['BRId',	"GenderId",	"IAId", "Amount of Credit Cards", "Nationality",	"Occupation",	"Fee Structure", "Loyalty Classification","Properties Owned"	, "Risk Weighting", "Income Band" ]]):
    plt.figure(i)
    sns.countplot(data=df, x=predictor, hue="Nationality")
```

```python
for col in categorical_cols:
  if col == "Occupation":
    continue
  plt.figure(figsize=(8,4) )
  sns.histplot(df[col])
  plt.title("Histogram plot of occupation count")
  plt.xlabel(col)
  plt.ylabel("count")
  plt.show()
```

Numerical analysis

```python
numerical_cols = ['Estimated Income','Superannuation Savings','Credit Card Balance','Bank Loans', 'Bank Deposits',	'Checking Accounts',	'Saving Accounts'	,'Foreign Currency Account','Business Lending']
#univariant analysis and visualization
plt.figure(figsize=(8,12))
for i , col in enumerate(numerical_cols):
  plt.subplot(4,3, i+1)
  sns.histplot(df[col], kde=True)
  plt.title(col)
plt.show()
```

```python
#Heatmap
numerical_cols = ['Estimated Income','Superannuation Savings','Credit Card Balance','Bank Loans', 'Bank Deposits',	'Checking Accounts',	'Saving Accounts'	,'Foreign Currency Account','Business Lending']
correlation_matrix = df[numerical_cols].corr()
plt.figure(figsize=(12, 12))
sns.heatmap(correlation_matrix, annot=True, cmap='crest', fmt=".2f")
plt.title("correlation matrix")
plt.show()
```

Insights of EDA
1. the strongest correction occurs among "bank deposit", "checking accounts", "savings acc", and "foriegn currency acc" indicating that customers who maintain high balances in one acc type often hold substainal other amounts as well.
